```python
import numpy as np
import pandas as pd
```

```python
retail = pd.read_csv('./assets/retail.csv')

mask = retail['family'].isin(['BABY CARE', 'AUTOMOTIVE', 'BEAUTY']) & retail['store_nbr'].isin(range(1,5))

small_retail = retail.loc[mask].sample(100, random_state=616)
small_retail.head()
```

```python
small_retail.pivot_table(columns='store_nbr',
                   index='family',
                   values='sales',
                   aggfunc='sum')
```

```python
small_retail.pivot_table(columns='store_nbr',
                   index='family',
                   values='sales',
                   aggfunc='sum',
                   margins=True)
```

```python
pm = pd.read_csv('./assets/premier_league_games.csv')
pm.head()
```

```python
(pm.query("season > '2015/2016' and home_team in ['Arsenal', 'Chelsea', 'Everton']")
    .pivot_table(columns='season',
               index='home_team',
               values='home_goals',
               aggfunc='sum',
                margins=True)
)
```

```python
small_retail.pivot_table(index='family',
                         columns='store_nbr',
                         values='sales',
                         aggfunc=('sum','mean'))
```

```python
premier_league = (pm.query("season > '2015/2016' and home_team in ['Arsenal', 'Chelsea', 'Everton']")
    .pivot_table(columns='season',
               index='home_team',
               values='home_goals',
               aggfunc='sum')
)
premier_league
```

```python
pm2 = premier_league.reset_index()
pm2
```
