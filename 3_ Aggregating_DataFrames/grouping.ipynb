```python
import numpy as np
import pandas as pd
```

```python
retail = pd.read_csv("./assets/retail.csv")
retail.groupby('family')
```

```python
retail.groupby('family')[['sales']].sum().head()
```

```python
premier_league = pd.read_csv('./assets/premier_league_games.csv')
premier_league.head()
```

```python
premier_league.groupby('home_team')[['home_goals']].mean().sort_values('home_goals', ascending=False).head()
```

```python
premier_league.groupby('away_team')[['away_goals']].mean().sort_values('away_goals', ascending=False).head()
```

```python
# multi index
retail.groupby(['family', 'store_nbr'])[['sales']].sum()
```

```python
retail.groupby(['family', 'store_nbr'], as_index=False)[['sales']].sum()
```

```python
(premier_league.groupby(['home_team', 'season'], as_index=False)[['home_goals']].sum()
    .sort_values(['home_team', 'season'], ascending=[True, False]))
```
