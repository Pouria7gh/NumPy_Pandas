```python
premier_league = pd.read_csv('./assets/premier_league_games.csv')
premier_league.head()
```

```python
pm = premier_league.assign(
    avg_home_team_home_goals = premier_league.groupby('home_team')['home_goals']
        .transform('mean'),
    difference = lambda x : x['home_goals'] - x['avg_home_team_home_goals']
)

pm.loc[pm['difference'] > 0]
```

```python
pm.groupby(['home_team', 'away_team']).agg({'difference': 'mean'}).loc['Arsenal'].sort_values('difference', ascending=False)
```

```python
pm.query("home_team == 'Arsenal' and away_team == 'Blackpool'")
```
