```python
import numpy as np
import pandas as pd
```

```python
retail = pd.read_csv('./assets/retail.csv')
retail.head()
```

```python
retail.groupby(['store_nbr', 'family'])['sales'].sum().round()
```

```python
cols = ['store_nbr', 'family', 'sales', 'onpromotion']
retail.loc[:,cols].groupby(['store_nbr', 'family']).agg('sum')
```

```python
retail.loc[:, cols].groupby(['store_nbr', 'family']).agg(['sum', 'mean'])
```

```python
(retail.groupby(['store_nbr', 'family'])
     .agg({'sales': ['sum', 'mean'],
           'onpromotion': ['min', 'max']})
)
```

```python
(retail.groupby(['store_nbr', 'family'])
     .agg(sales_mean=('sales', 'mean'),
          onpromotion_max=('onpromotion', 'max'))
)
```

```python
premier_league = pd.read_csv('./assets/premier_league_games.csv')
premier_league.head()
```

```python
useless_cols = ['league_name', 'away_team', 'id']

(premier_league.drop(useless_cols,axis = 1)
    .groupby(['season', 'home_team'], as_index=False)
    .agg('sum')
)
```

```python
(premier_league.drop(useless_cols, axis=1)
     .groupby(['season', 'home_team'], as_index=False)
     .agg(['sum', 'mean'])
)
```

```python
(premier_league.groupby(['season', 'home_team'],as_index=False)
 .agg({'home_goals': ['count', 'sum'],
       'away_goals': ['mean', 'max']})
)
```

```python
(premier_league.groupby(['season', 'home_team'],as_index=False)
 .agg(home_goal_sum=('home_goals', 'sum'),
      home_goal_mean=('home_goals', 'mean'))
)
```
