```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('./assets/retail.csv', parse_dates=['date'])
df.head()
```

```python
sales_df = (df.groupby(['date', 'family'], as_index=False)
    .agg(sales=('sales', 'sum'))
    .pivot(index='date', columns='family', values='sales')
)
```

```python
(sales_df
    .loc["2016-6", ["BEVERAGES","BEAUTY","BABY CARE","AUTOMOTIVE"]]
    .plot(
        subplots=True,
        layout=(2,2),
        sharey= True,
        figsize=(8, 8)
    )
);
```
