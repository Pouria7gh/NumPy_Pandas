```python
import numpy as np
import pandas as pd
```

```python
clean_wholesale = pd.Series(
    [3.99, 5.99, 22.99, 7.99, 33.99, 10]
)
clean_wholesale
```

```python
clean_wholesale.where(clean_wholesale <= 20, round(clean_wholesale * .9, 2))
```

```python
clean_wholesale.where(~(clean_wholesale > 20), round(clean_wholesale*.9, 2))
```

```python
clean_wholesale.where(
    ~(clean_wholesale > 20),
    round(clean_wholesale*.9, 2)
).where(
    clean_wholesale.gt(10),
    0
)
```

```python
np.where(clean_wholesale > 20, round(clean_wholesale*.9, 2),
         np.where(clean_wholesale <= 10, 0, clean_wholesale))
```
