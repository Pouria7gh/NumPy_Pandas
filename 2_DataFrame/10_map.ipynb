```python
import numpy as np
import pandas as pd
```

```python
product_df = pd.DataFrame(
    {'product_id': range(5),
     'product': ['dairy', 'dairy', 'dairy', 'vegetables', 'fruits'],
     'price': [2.56, 2.56, 4.55, 2.74, 5.44]}
)
product_df
```

```python
mapping_dict = {'dairy': 'non_vegan', 'vegetables': 'vegan', 'fruits': 'vegan'}
product_df['vegan?'] = product_df['product'].map(mapping_dict)
product_df
```

```python
product_map = {'dairy': 0, 'vegetables': 1, 'fruits': 2}
vegan_map = {'non_vegan': 0, 'vegan': 1}
product_df['product'] = product_df['product'].map(product_map).astype('int')
product_df['vegan?'] = product_df['vegan?'].map(vegan_map).astype('int')
product_df
```

```python
product_df['price'] = product_df['price'].map(lambda x: f"${x}")
product_df
```

```python
product_df['price'] = product_df['price'].apply(lambda x: f"{x} dollars")
product_df
```

```python
product_df['price'] = product_df['price'].map(lambda x: x[1:5])
product_df
```
