```python
import numpy as np
import pandas as pd
```

```python
retail_df = pd.read_csv('./assets/retail.csv')
condition = retail_df['family'].isin(['BEVERAGES', 'DELL', 'DAIRY'])
sample_df = retail_df.loc[condition].sample(5, random_state=2021)
sample_df
```

```python
sample_df.sort_index(axis=1)
```

```python
sample_df.sort_index(axis=1, ascending=False)
```

```python
sample_df.sort_index(axis=1, ascending=True, inplace=True)
sample_df
```

```python
sample_df.sort_values('store_nbr')
```

```python
sample_df.sort_values('store_nbr', ascending=False)
```

```python
sample_df.sort_values(['family', 'sales'], ascending=[True, False])
```
