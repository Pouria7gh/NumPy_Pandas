```python
import numpy as np
import pandas as pd
```

```python
retail_df = pd.read_csv('./assets/retail.csv')
retail_df.head()
```

```python
condition = retail_df['family'].isin(['BABY CARE', 'BOOKS']) & retail_df['sales'] > 0
baby_books = retail_df.loc[condition].sample(5, random_state=2018)
baby_books
```

```python
baby_books['tax_amount'] = baby_books['sales'] * 0.05
baby_books['total'] = baby_books['sales'] + baby_books['tax_amount']
baby_books
```

```python
baby_books['taxable_category'] = baby_books['family'] != 'BABY CARE'
baby_books.drop(['tax_amount', 'total'], axis=1, inplace=True)
baby_books
```

```python
baby_books.drop('taxable_category', axis=1, inplace=True)
baby_books['tax_amount'] = (
    baby_books['sales'] * 0.05 * (baby_books['family'] != 'BABY CARE')
)
baby_books
```
