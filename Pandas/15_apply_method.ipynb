```python
import numpy as np
import pandas as pd
```

```python
clean_wholesale = pd.Series(
    [3.99, 5.99, 22.99, 7.99, 33.99]
)
clean_wholesale
```

```python
def discount(price):
    if price > 20:
        return round(price * 0.9, 2)
    return price
```

```python
clean_wholesale.apply(discount)
```

```python
clean_wholesale.apply(lambda x: round(x*0.9, 2) if x > 20 else x)
```

```python
clean_wholesale.where(
    clean_wholesale <= 20, round(clean_wholesale*0.9, 2))
```

```python
string_series = pd.Series(
    ['day 0', 'day 0', 'day 2', 'day 2', 'day 4']
)
string_series
```

```python
string_series.apply(lambda x : x[-1])
```

```python
string_series.str.strip().str[-1]
```

```python
def search(string, looking_for):
    if looking_for in string:
        return 'Found it!'
    return 'Nope'
```

```python
string_series.apply(lambda x: search(x, '4'))
```

```python
search = '0'
string_series.where(string_series.str.contains(search), "Nope").where(~string_series.str.contains(search), "Found it!")
```

```python
np.where(string_series.str.contains(search), 'Found it!', 'Nop')
```
