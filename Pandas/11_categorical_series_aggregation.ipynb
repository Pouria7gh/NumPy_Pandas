```python
import pandas as pd
import numpy as np
```

```python
items = pd.Series(['coffee', 'coffee', 'tea', 'coconut', 'sugar',])
items
```

```python
items.value_counts()
```

```python
items.value_counts(normalize=True)
```

```python
'%' + items.value_counts(normalize=True).mul(100).round(2).astype('string')
```

```python
items.unique()
```

```python
items.nunique()
```
